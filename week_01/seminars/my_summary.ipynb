series - базовая структура данных, одномерный массив с метками (индексами)

Создание
`pandas.Series(data=None, index=None, dtype=None, name=None, copy=None)`
- **data** - источник данных
- **index** - Метки для каждого элемента (по умолчанию числовые индексы от 0)
- **dtype** - тип данных массива
- **name** - имя для series (полузно при составлении столбцов в таблице)
- **copy** - если True то данные копируются, по умолчанию False

````python
data = [1, 2, 13, -17, 90, 34]
series = pd.Series(data)
series
````

````python
data = [1, -2, 3]
coord = pd.Series(data, index=['X', 'Y', 'Z'])
coord
````

# Библиотека matplotlib и numpy

````python
import matplotlib.pyplot as plt
import numpy as np
````

````python
# Задаем значения осей
x = np.arange(10)
y = np.random.randint(1, 10, size=10)
x, y
````

````python
# Задаем размер полотна
plt.figure(figsize=(10,7))

# Построим линейную диаграмму
plt.plot(x, y)

# Добавим название
plt.title('График зависимости x от y')

# Добавим подписи к осям
plt.xlabel('X')
plt.ylabel('Y')

# нарисуем
plt.show()
````

изменим график на scatter (точки не соединенные линиями)

````python
# Задаем размер полотна
plt.figure(figsize=(7,10))

# Построим линейную диаграмму
plt.scatter(x, y)

# Добавим название
plt.title('График зависимости x от y')

# Добавим подписи к осям
plt.xlabel('X')
plt.ylabel('Y')

# нарисуем
plt.show()
````

Добавим цвета

````python
x = np.arange(10)
y = np.random.randint(1, 10, size=10)
z = np.random.randint(1, 10, size=10)

plt.figure(figsize=(10, 7))

plt.plot(x, y, color="#68904D", label='данные y')
plt.plot(x, z, color='#EE9B01', label='данные z')

plt.legend(fontsize=15)

plt.title('Данные vs время', fontsize=20)

plt.xlabel("Время", fontsize=20)
plt.ylabel("Данные", fontsize=20)

# Можно ограничить координаты осей
# plt.xlim([-5, 20])
# plt.ylim([-5, 20])

# нарисовать сетку
plt.grid()

plt.show()
````

Метод dropna - очищает пустые данные (удаляет любую строку, где есть хотя бы 1 NaN)

- axis - удалять строки или столбцы
- how - 'any' удалит ьесли хотя бы 1, 'all' - удалить если все NaN
- thresh=count - оставить строку, если в ней заполнено только count значений
- subset=['name_col_1', 'name_col_2', ...] - ищет пропуски только в указанных столбцах
- inplace Если True то изменения применятся к текущей таблицу, иначе вернется новая (по умол. False) 

````python
data = df['Age'].dropna()
mean_Age = data.mean()
mean_Age
````

````python
median_Age = data.quantile(0.5)
median_Age
````

Нарисуем распределение признака Age с его средним и медианой

````python
plt.hist(df['Age'])

plt.axvline(x=mean_Age, color='r', linestyle='-', label='mean')
plt.axvline(x=median_Age, color='g', linestyle='--', label='median')

plt.show()
````
